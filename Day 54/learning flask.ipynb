### Python DEcorator function

```python
import time

#decorative functions
def delay_decorator(function):
    def wrapper_function():
        time.sleep(2)
        #Do somthing before running the function
        function()
        function()
        #do something after running the function
    return wrapper_function

@delay_decorator
def say_hello():
    print("Hello")

@delay_decorator
def say_bye():
    print("Bye")


def say_greeting():
    print("How are you?")


say_greeting()
say_hello()

decorated_function=delay_decorator(say_greeting)
decorated_function()
```
